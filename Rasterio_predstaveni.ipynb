## Co je `rasterio` a k čemu slouží?

**`rasterio`** je open-source knihovna v Pythonu určená pro čtení, zápis, transformace a analýzu rastrů (např. formát GeoTIFF). Uvnitř staví na knihovně **GDAL**, ale místo jejího nízkoúrovňového rozhraní nabízí čisté, intuitivní Python API a přímou vazbu na **NumPy**.

Hlavní funkce a možnosti knihovny `rasterio`:
* **Práce s metadaty a georeferencí:** čtení souřadnicového systému (`crs`), geografického rozsahu (`bounds`), rozlišení (`res`) a afinní transformační matice (`transform`).
* **Načítání dat jako pole NumPy:** rastrová pásma lze načítat jako vícerozměrná pole a provádět nad nimi rychlou rastrovou algebru.
* **Práce s výřezy (`Window`):** umožňuje načíst jen vybrané okno rastru bez nutnosti zpracovávat celý rozsáhlý snímek.
* **Převzorkování a reprojekce (`reproject`):** sjednocuje rastry s různým prostorovým rozlišením nebo souřadnicovým systémem na společnou mřížku.
* **Maskování oblačnosti:** umožňuje omezit výpočty na zájmové pixely či vrstvy.

Samotné `rasterio` neslouží k pokročilé kartografické kompozici nebo statistickým grafům – proto se skvěle kombinuje s knihovnami **NumPy** (výpočty a indexy) a **Matplotlib** (vizualizace).

```python
!pip install -q rasterio pystac-client planetary-computer matplotlib numpy

import numpy as np
import rasterio
from rasterio.windows import from_bounds
from rasterio.warp import transform_bounds, reproject, Resampling
import pystac_client
import planetary_computer as pc
import matplotlib.pyplot as plt

print("Verze rasterio:", rasterio.__version__)
```

## 1. Vyhledání scény a metadat v `rasterio`

Vyhledáme scénu Sentinel-2 L2A pro severní část Prahy přes STAC a přes `rasterio.open()` zkontrolujeme základní prostorové vlastnosti rastru.

```python
bbox_wgs84 = [14.402, 50.098, 14.432, 50.118]

catalog = pystac_client.Client.open(
    "https://planetarycomputer.microsoft.com/api/stac/v1",
    modifier=pc.sign_inplace,
)

search = catalog.search(
    collections=["sentinel-2-l2a"],
    bbox=bbox_wgs84,
    datetime="2023-05-01/2023-08-31",
    query={"eo:cloud_cover": {"lt": 15}},
)

items = list(search.items())
item = items[0]

print(f"Nalezeno scén: {len(items)} | Vybrána: {item.id} ({item.datetime})")

with rasterio.open(item.assets["B08"].href) as src:
    print("Souřadnicový systém (CRS):", src.crs)
    print("Rozsah (bounds):          ", src.bounds)
    print(f"Rozměry rastru (v × š):   {src.height} × {src.width} px")
    print("Velikost pixelu (res):    ", src.res)
    print("Datový typ:               ", src.dtypes[0])
    print("\nAfinní transformace:\n", src.transform)
```

## 2. Výřez přes `Window` a kalibrace

Pomocí `Window` načteme jen vybraný výřez 10m pásem a 20m vrstvy `SCL` a digitální hodnoty převedeme na odrazivost.

```python
def nacti_okno(asset_key, bbox_4326):
    href = item.assets[asset_key].href
    with rasterio.open(href) as src:
        bbox_crs = transform_bounds("EPSG:4326", src.crs, *bbox_4326)
        win = from_bounds(*bbox_crs, transform=src.transform).round_offsets().round_lengths()
        arr = src.read(1, window=win)
        win_transform = src.window_transform(win)
        crs = src.crs
        res = src.res
    return arr, win_transform, crs, res

b02_raw, win_transform_10m, crs, res_10m = nacti_okno("B02", bbox_wgs84)
b03_raw, _, _, _ = nacti_okno("B03", bbox_wgs84)
b04_raw, _, _, _ = nacti_okno("B04", bbox_wgs84)
b08_raw, _, _, _ = nacti_okno("B08", bbox_wgs84)
scl_20m, win_transform_20m, _, res_20m = nacti_okno("SCL", bbox_wgs84)

def kalibruj(pasmo):
    return (pasmo.astype("float32") - 1000.0) / 10000.0

b02, b03, b04, b08 = kalibruj(b02_raw), kalibruj(b03_raw), kalibruj(b04_raw), kalibruj(b08_raw)
print(f"10m výřez: {b08.shape} ({res_10m}) | 20m SCL výřez: {scl_20m.shape} ({res_20m})")
```

## 3. Sjednocení rozlišení přes `reproject`

Vrstvu `SCL` (20 m) převedeme funkcí `reproject()` s metodou `Resampling.nearest` na 10m mřížku.

```python
scl = np.empty(b08.shape, dtype="int16")

reproject(
    source=scl_20m.astype("int16"),
    destination=scl,
    src_transform=win_transform_20m,
    src_crs=crs,
    dst_transform=win_transform_10m,
    dst_crs=crs,
    resampling=Resampling.nearest,
)

print("Rozměr SCL po sjednocení:", scl.shape, "| unikátní kódy:", np.unique(scl))
```

## 4. Kompozice True Color a False Color

Porovnáme přirozené barvy (`B04, B03, B02`) a nepravou barevnou kompozici (`B08, B04, B03`).

```python
def kompozice(r, g, b, max_refl=0.35):
    return np.clip(np.dstack((r, g, b)) / max_refl, 0.0, 1.0)

rgb_vis = kompozice(b04, b03, b02)
cir_vis = kompozice(b08, b04, b03)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
ax1.imshow(rgb_vis)
ax1.set_title("True Color (B04 – B03 – B02)")
ax1.axis("off")

ax2.imshow(cir_vis)
ax2.set_title("False Color / CIR (B08 – B04 – B03)")
ax2.axis("off")

plt.tight_layout()
plt.show()
```

## 5. NDVI a porovnání povrchů

V polích NumPy spočítáme **NDVI** po vyřazení neplatných pixelů (`SCL` kódy `0, 1, 3, 8, 9, 10`) a porovnáme vegetaci (4) s vodou (6).

```python
platne = ~np.isin(scl, [0, 1, 3, 8, 9, 10])
jmenovatel = b08 + b04

with np.errstate(divide="ignore", invalid="ignore"):
    ndvi = np.where(platne & (jmenovatel != 0), (b08 - b04) / jmenovatel, np.nan)

vegetace = scl == 4
voda = scl == 6

print(f"Průměrné NDVI – vegetace: {np.nanmean(ndvi[vegetace]):.3f} | voda: {np.nanmean(ndvi[voda]):.3f}")

fig, ax = plt.subplots(figsize=(8, 5.5))
im = ax.imshow(ndvi, cmap="RdYlGn", vmin=-0.5, vmax=0.8)
plt.colorbar(im, ax=ax, label="NDVI")
ax.set_title("Index NDVI po maskování neplatných pixelů (SCL)")
ax.axis("off")
plt.show()
```

## 6. Maska vegetace

Z vypočteného `NDVI` a platných pixelů vytvoříme binární masku vegetace (např. práh `NDVI > 0.3` v rámci platných pixelů, případně v kombinaci s kódem `SCL == 4`) a zobrazíme ji.

```python
maska_vegetace = platne & (ndvi > 0.3)

print("Počet pixelů vegetace (NDVI > 0.3):", int(np.sum(maska_vegetace)))

fig, ax = plt.subplots(figsize=(8, 5.5))
ax.imshow(maska_vegetace, cmap="Greens")
ax.set_title("Maska vegetace (platné pixely s NDVI > 0.3)")
ax.axis("off")
plt.show()
```
